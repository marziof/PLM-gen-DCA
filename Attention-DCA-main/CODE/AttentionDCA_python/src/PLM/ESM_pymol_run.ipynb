```python
filename="rmsd_matrix.npy"
np.save(filename, rmsd_results)
rmsd_results = np.load(filename,allow_pickle=True).item()
print(rmsd_results)
def dict_to_matrix(rmsd_dict):
    # Find the size of the matrix
    max_i = max(key[0] for key in rmsd_dict.keys())
    max_j = max(key[1] for key in rmsd_dict.keys())
    size = max(max_i, max_j) + 1  # Add 1 since indexing starts at 0

    # Initialize the matrix with NaNs or zeros (choose depending on use case)
    matrix = np.full((size, size), np.nan)

    # Fill the matrix
    for (i, j), value in rmsd_dict.items():
        matrix[i, j] = value

    return matrix
rmsd_matrix = dict_to_matrix(rmsd_results)

print(rmsd_matrix)
```
